```python
def vacuum_cleaner(vacuum_location,
                   room_a_dirt,
                   room_b_dirt,
                   room_a_obstacle,
                   room_b_obstacle,
                   battery):
    """
    Vacuum Cleaner Agent

    vacuum_location : 'A' or 'B'
    room_a_dirt     : percentage of dirt in Room A (0-100)
    room_b_dirt     : percentage of dirt in Room B (0-100)
    room_a_obstacle : percentage of Room A blocked by obstacles (0-100)
    room_b_obstacle : percentage of Room B blocked by obstacles (0-100)
    battery         : battery percentage (0-100)

    The vacuum cleans only the area that is not blocked by obstacles.
    """

    print("\n==============================")
    print("      INITIAL STATE")
    print("==============================")

    print(f"Vacuum is in Room: {vacuum_location}")
    print(f"Battery: {battery}%")

    print(f"\nRoom A:")
    print(f"  Dirt: {room_a_dirt}%")
    print(f"  Obstacle: {room_a_obstacle}%")
    print(f"  Cleaned: {100 - room_a_dirt}%")

    print(f"\nRoom B:")
    print(f"  Dirt: {room_b_dirt}%")
    print(f"  Obstacle: {room_b_obstacle}%")
    print(f"  Cleaned: {100 - room_b_dirt}%")

    print("==============================")

    # Keep working while battery is available
    while battery > 0:

        # =====================================================
        # VACUUM IS IN ROOM A
        # =====================================================
        if vacuum_location == 'A':

            # Check if Room A has dirt
            if room_a_dirt > 0:

                # Calculate area that can actually be cleaned
                cleanable_area = 100 - room_a_obstacle

                # How much dirt can be cleaned
                cleaning_amount = min(room_a_dirt, cleanable_area)

                print("\nVacuum is in Room A.")
                print(f"Room A has {room_a_dirt}% dirt.")
                print(f"Room A has {room_a_obstacle}% obstacles.")
                print(f"Cleanable area: {cleanable_area}%")

                # Cleaning uses battery
                battery -= 10

                if battery < 0:
                    battery = 0

                # Remove dirt
                room_a_dirt -= cleaning_amount

                print(f"Action: Cleaning Room A...")
                print(f"Cleaned {cleaning_amount}% of Room A.")
                print(f"Battery decreased to {battery}%.")

                # If battery is empty, stop
                if battery == 0:
                    print("\nBattery is empty!")
                    break

            else:
                # Room A is already clean
                print("\nRoom A is already clean.")
                print("Moving to Room B...")

                battery -= 5

                if battery < 0:
                    battery = 0

                vacuum_location = 'B'

                print(f"Battery decreased to {battery}%.")

        # =====================================================
        # VACUUM IS IN ROOM B
        # =====================================================
        elif vacuum_location == 'B':

            # Check if Room B has dirt
            if room_b_dirt > 0:

                # Calculate area that can actually be cleaned
                cleanable_area = 100 - room_b_obstacle

                # How much dirt can be cleaned
                cleaning_amount = min(room_b_dirt, cleanable_area)

                print("\nVacuum is in Room B.")
                print(f"Room B has {room_b_dirt}% dirt.")
                print(f"Room B has {room_b_obstacle}% obstacles.")
                print(f"Cleanable area: {cleanable_area}%")

                # Cleaning uses battery
                battery -= 10

                if battery < 0:
                    battery = 0

                # Remove dirt
                room_b_dirt -= cleaning_amount

                print(f"Action: Cleaning Room B...")
                print(f"Cleaned {cleaning_amount}% of Room B.")
                print(f"Battery decreased to {battery}%.")

                # If battery is empty, stop
                if battery == 0:
                    print("\nBattery is empty!")
                    break

            else:
                # Room B is already clean
                print("\nRoom B is already clean.")
                print("Moving to Room A...")

                battery -= 5

                if battery < 0:
                    battery = 0

                vacuum_location = 'A'

                print(f"Battery decreased to {battery}%.")

        # =====================================================
        # CHECK WHETHER BOTH ROOMS ARE CLEAN
        # =====================================================

        # Maximum dirt that can remain because of obstacles
        room_a_unavoidable_dirt = room_a_obstacle
        room_b_unavoidable_dirt = room_b_obstacle

        # If remaining dirt is only under obstacles,
        # the vacuum has done all possible cleaning.
        if room_a_dirt <= room_a_unavoidable_dirt and \
           room_b_dirt <= room_b_unavoidable_dirt:

            print("\nAll accessible areas have been cleaned!")
            break

    # =========================================================
    # FINAL STATE
    # =========================================================

    print("\n==============================")
    print("        FINAL STATE")
    print("==============================")

    print(f"Vacuum is in Room: {vacuum_location}")
    print(f"Battery remaining: {battery}%")

    # Calculate cleaning percentages
    room_a_cleaned = 100 - room_a_dirt
    room_b_cleaned = 100 - room_b_dirt

    print("\nRoom A:")
    print(f"  Obstacle: {room_a_obstacle}%")
    print(f"  Dirt remaining: {room_a_dirt}%")
    print(f"  Cleaning percentage: {room_a_cleaned}%")

    print("\nRoom B:")
    print(f"  Obstacle: {room_b_obstacle}%")
    print(f"  Dirt remaining: {room_b_dirt}%")
    print(f"  Cleaning percentage: {room_b_cleaned}%")

    print("==============================")
```
